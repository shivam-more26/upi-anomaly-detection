# 04. Evaluation, PCA & Synthetic Sanity Checking

This notebook demonstrates:
- PCA 2D scatter visualization of normal vs anomaly points
- Anomaly explanation signal distribution
- Post-hoc synthetic truth sanity checking (strictly labeled as post-hoc validation, never model accuracy).

```python
import sys
from pathlib import Path
PROJECT_ROOT = Path('.').resolve().parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.utils import load_scored_data

df_scored = load_scored_data()
print(f"Loaded scored dataset: {len(df_scored)} records.")
```

## 1. PCA 2D Scatter Plot
Showing spatial separation of top percentile high-risk points.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df_scored,
    x='pca_x',
    y='pca_y',
    hue='risk_level',
    palette={'HIGH_RISK': 'red', 'SUSPICIOUS': 'orange', 'NORMAL': 'lightgrey'},
    alpha=0.7,
    s=25
)
plt.title('2D PCA Projection of Features (Normal vs High-Risk Points)')
plt.xlabel('PCA Dimension 1')
plt.ylabel('PCA Dimension 2')
plt.legend(title='Risk Level')
plt.tight_layout()
plt.show()
```

## 2. Post-Hoc Synthetic Anomaly Sanity Check
Comparing top percentile flagged risk categories with injected synthetic anomaly labels.
**NOTE:** This post-hoc evaluation is for sanity checking synthetic data patterns only, NOT a real-world accuracy score.

```python
ct = pd.crosstab(df_scored['risk_level'], df_scored['synthetic_anomaly_truth'], margins=True)
ct.columns = ['Normal Synthetic', 'Injected Anomaly', 'Total']
ct
```
